```python
inventario = {
    "SKU-01": {"nombre": "Teclado RGB", "stock_actual": 16, "stock_minimo": 5, "costo_unitario": 40.0},
    "SKU-02": {"nombre": "Monitor 24", "stock_actual": 4, "stock_minimo": 5, "costo_unitario": 150.0},
    "SKU-03": {"nombre": "Mouse Gamer", "stock_actual": 8, "stock_minimo": 10, "costo_unitario": 25.0},
    "SKU-04": {"nombre": "Audifonos Pro", "stock_actual": 2, "stock_minimo": 4, "costo_unitario": 65.0},
}

transacciones = [
    {"tipo": "venta", "sku": "SKU-01", "cantidad": 12},
    {"tipo": "venta", "sku": "SKU-02", "cantidad": 6},
    {"tipo": "devolucion", "sku": "SKU-03", "cantidad": 3},
    {"tipo": "venta", "sku": "SKU-04", "cantidad": 2},
    {"tipo": "venta", "sku": "SKU-99", "cantidad": 1},
]

incidencias = []
print("Datos listos.")
```

```python
for tx in transacciones:
    tipo = tx["tipo"]
    sku = tx["sku"]
    cantidad = tx["cantidad"]

    if sku not in inventario:
        incidencias.append("SKU no encontrado: " + sku)
    else:
        if tipo == "venta":
            disponible = inventario[sku]["stock_actual"]
            if disponible >= cantidad:
                inventario[sku]["stock_actual"] = disponible - cantidad
            else:
                incidencias.append("Stock insuficiente en " + sku)
        elif tipo == "devolucion":
            inventario[sku]["stock_actual"] = inventario[sku]["stock_actual"] + cantidad
```

```python
valor_total = 0

for sku, datos in inventario.items():
    valor_item = datos["stock_actual"] * datos["costo_unitario"]
    valor_total += valor_item

print(f"Valor total del inventario: {valor_total:.2f}")
```

```python
alertas_reposicion = []

for sku, datos in inventario.items():
    if datos["stock_actual"] < datos["stock_minimo"]:
        alertas_reposicion.append(sku + " (" + datos["nombre"] + ")")
```

```python
print("Incidencias:", incidencias)
print("Valor total del inventario:", valor_total)
print("Alertas de reposición:", alertas_reposicion)
```
